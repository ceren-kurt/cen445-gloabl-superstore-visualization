```python
bins = [-0.01, 0.0, 0.20, 0.40, 1.0]
labels = ["0%", "1%–20%", "21%–40%", "41%+"]

df["discount_group"] = pd.cut(df["discount"], bins=bins, labels=labels)
```

```python
#check the outliers

q1, q3 = df["profit"].quantile([.25, .75])
iqr = q3 - q1
df["profit_outlier"] = (df["profit"] < q1 - 1.5*iqr) | (df["profit"] > q3 + 1.5*iqr)
```

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(f"Lower Bound: {lower_bound:.2f} | upper bound: {upper_bound:.2f}")
print("Total outlier num:", df["profit_outlier"].sum())
print(f"ratio in dataaset: %{(df['profit_outlier'].mean() * 100):.2f}")

# En uç zarar ve en uç kâr eden ilk 10 outlier satırı görelim
outliers = df[df["profit_outlier"]].sort_values(by="profit")
display(outliers[["order_date", "market", "category", "sub_category", "sales", "discount", "profit"]].head(5))
display(outliers[["order_date", "market", "category", "sub_category", "sales", "discount", "profit"]].tail(5))
```
